```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from pathlib import Path
```

```python
warnings.filterwarnings('ignore')
```

```python
# Set visualization styles
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
```

```python
# Path definitions
DATA_DIR = Path("../student_resource/dataset/train")


train_s1 = pd.read_csv(DATA_DIR / "train_source1.tsv", sep="\t")
train_s2 = pd.read_csv(DATA_DIR / "train_source2.tsv", sep="\t")
train_s3 = pd.read_csv(DATA_DIR / "train_source3.tsv", sep="\t")
train_gt = pd.read_csv(DATA_DIR / "train_ground_truth.tsv", sep="\t")
```

```python
print(f"Source 1 Records: {len(train_s1):,}")
print(f"Source 2 Records: {len(train_s2):,}")
print(f"Source 3 Records: {len(train_s3):,}")
print(f"Ground Truth Rows: {len(train_gt):,}")
```

```python
# 1. Fill NaNs in ground truth (NaN means no matches / singleton)
train_gt['matched_entity_ids'] = train_gt['matched_entity_ids'].fillna('')

# 2. Count number of target matches per S1 entity
train_gt['match_count'] = train_gt['matched_entity_ids'].apply(
    lambda x: 0 if x == '' else len([m for m in x.split(',') if m.strip()])
)

# 3. Calculate exact ratios
total_s1 = len(train_gt)
singletons = (train_gt['match_count'] == 0).sum()
singleton_ratio = (singletons / total_s1) * 100

print(f"=== GROUND TRUTH METRICS ===")
print(f"Total S1 Entities : {total_s1:,}")
print(f"Singletons (0 matches) : {singletons:,} ({singleton_ratio:.2f}%)")
print(f"Non-Singletons (>0)   : {total_s1 - singletons:,} ({100 - singleton_ratio:.2f}%)\n")

# 4. Check for missing values in input data
print("=== MISSING VALUE CHECK ===")
for name, df in [("Source 1", train_s1), ("Source 2", train_s2), ("Source 3", train_s3)]:
    nulls = df[['business_name', 'business_address', 'country']].isnull().sum()
    print(f"--- {name} Null Counts ---")
    print(nulls)
    print()
```
